Import Libraries

```python
import numpy as np
from datetime import datetime
```

Read the Dataset

```python
with open("hostel_bois.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()

print("Total Lines:", len(lines))
```

Initialize Variables

```python
messages = []

system_count = 0
media_count = 0
deleted_count = 0
```

Parse the Chat (Part 1)

```python
system_messages = []

for line in lines:
    line = line.strip()

    if line == "":
        continue

    if " - " not in line:
        continue

    timestamp, rest = line.split(" - ", 1)

    if ": " not in rest:
        system_messages.append(line)

print("System Messages:")
for msg in system_messages:
    print(msg)
```

```python
print("=" * 40)
print("PARSER SUMMARY")
print("=" * 40)
print("Total Lines      :", len(lines))
print("Valid Messages   :", len(messages))
print("System Messages  :", system_count)
print("Media Messages   :", media_count)
print("Deleted Messages :", deleted_count)
```

Verify the Parser

```python
print("Total Lines:", len(lines))
print("Valid Messages:", len(messages))
print("Skipped/System Messages:", len(lines) - len(messages))
```

# Feature 2 : Group Overview

Total Messages

```python
total_messages = len(messages)

print("Total Messages :", total_messages)
```

Total Participants

```python
participants = set()

for msg in messages:
    participants.add(msg["sender"])

print("Total Participants :", len(participants))
print("Participants :", participants)
```

```python
message_count = {}

for msg in messages:
    sender = msg["sender"]

    if sender in message_count:
        message_count[sender] += 1
    else:
        message_count[sender] = 1
```

Messages Per Person

```python
print("=" * 40)
print("Messages Sent by Each Participant")
print("=" * 40)

sorted_messages = sorted(message_count.items(), key=lambda x: x[1], reverse=True)

for person, count in sorted_messages:
    print(f"{person:<10} : {count}")
```

# Feature 3 : Most Active Day

```python
day_count = {}

for msg in messages:
    date = msg["timestamp"].split(",")[0]

    if date in day_count:
        day_count[date] += 1
    else:
        day_count[date] = 1
```

```python
most_active_day = max(day_count, key=day_count.get)

print("Most Active Day :", most_active_day)
print("Messages :", day_count[most_active_day])
```

# Feature 4 : Most Active Hour

```python
hour_count = {}

for msg in messages:

    time = msg["timestamp"].split(",")[1].strip()

    hour = int(time.split(":")[0])

    if hour in hour_count:
        hour_count[hour] += 1
    else:
        hour_count[hour] = 1

most_active_hour = max(hour_count, key=hour_count.get)

print("Most Active Hour :", most_active_hour)
print("Messages :", hour_count[most_active_hour])
```

# Feature 5 : Activity Heatmap

Create NumPy Matrix

```python
import numpy as np
from datetime import datetime
```

```python
import numpy as np

print(np.__version__)
```

```python
participants = ["Rahul", "Priya", "Neha", "Aman", "Karan", "Vikas"]

person_index = {
    "Rahul": 0,
    "Priya": 1,
    "Neha": 2,
    "Aman": 3,
    "Karan": 4,
    "Vikas": 5
}

heatmap = np.zeros((6, 24), dtype=int)
```

```python
for msg in messages:

    sender = msg["sender"]

    hour = int(msg["timestamp"].split(",")[1].strip().split(":")[0])

    row = person_index[sender]

    heatmap[row][hour] += 1
```

```python
print("Heatmap Matrix:\n")
print(heatmap)
```

```python
print("=" * 95)
print("ACTIVITY HEATMAP (Hours 00-23)")
print("=" * 95)

# Print hour headings
print("Participant".ljust(12), end=" ")
for hour in range(24):
    print(f"{hour:02}", end=" ")
print()

print("-" * 95)

# Print each participant's activity
for i, person in enumerate(participants):

    print(person.ljust(12), end=" ")

    for value in heatmap[i]:

        if value == 0:
            symbol = "."
        elif value <= 10:
            symbol = "░"
        elif value <= 30:
            symbol = "▒"
        else:
            symbol = "█"

        print(symbol, end="  ")

    print()
```

# Feature 6 : Top Words

```python
word_count = {}

for msg in messages:

    words = msg["message"].lower().split()

    for word in words:

        word = word.strip(".,!?()[]{}\"'")

        if len(word) < 2:
            continue

        if word in word_count:
            word_count[word] += 1
        else:
            word_count[word] = 1
```

```python
sorted_words = sorted(word_count.items(), key=lambda x: x[1], reverse=True)

print("=" * 40)
print("Top 10 Words")
print("=" * 40)

for word, count in sorted_words[:10]:
    print(f"{word:<15} {count}")
```

```python
most_active = max(message_count, key=message_count.get)
print("Most Active Member :", most_active)
print("Messages Sent :", message_count[most_active])
```
